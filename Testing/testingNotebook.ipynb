**Test 2: Pyastronomy**

The aim of this test is to determine if the cross correlation function is accurate. For this test, I will use pyastronomy, which is a package that already has a cross correlation method built in. 

```python
import sys
print(sys.executable)
!{sys.executable} -m pip install PyAstronomy
#AI said this is how I can use files that appear in the parent directory
import os
parent_dir = os.path.dirname(os.getcwd())
sys.path.append(parent_dir)
import data_prep as dp
import ccf
from radialVelocityCalc import getHeaderValues
from PyAstronomy import pyasl
import numpy as np

#This should allow the files in the current git library to be used as inputs. So you should be able to run this test from your computer as well
testing_dir = os.getcwd()

#First, I am testing on two spectra that the LCO pipeline succeeded on. The LCO results for these files are also in the git repo
observedFile = os.path.join(testing_dir, "testingGoodObserved.fits.fz")
templateFile = os.path.join(testing_dir, "testingGoodTemplate.fits.fz")
```

```python
#gets header values
barycorrObs, barycorrTemp, templateRV = getHeaderValues(observedFile, templateFile)
#same setup as 
obsWavelengths, obsFluxes = dp.readingAndReadyingData(observedFile)
tempWavelengths, tempFluxes = dp.readingAndReadyingData(templateFile)
obsWavelengths, obsFluxes, tempWavelengths, tempFluxes = dp.makeOverlap(obsWavelengths, obsFluxes, tempWavelengths, tempFluxes)
#newWavelengths, newObsFluxes, newTempFluxes = dp.interpolation(obsWavelengths, obsFluxes, tempWavelengths, tempFluxes)

rvList, ccfList = pyasl.crosscorrRV(obsWavelengths, obsFluxes, tempWavelengths, tempFluxes, rvmin=-200, rvmax=200, drv =1.0,skipedge=200)

rv = rvList[np.argmax(ccfList)]*1000 +templateRV + (barycorrObs - barycorrTemp)

print("Rv: "+ str(rv))
```
