**Test 2: Pyastronomy**

The aim of this test is to determine if the cross correlation function is accurate. For this test, I will use pyastronomy, which is a package that already has a cross correlation method built in. 

```python
import sys
print(sys.executable)
!{sys.executable} -m pip install PyAstronomy
#AI said this is how I can use files that appear in the parent directory
import os
parent_dir = os.path.dirname(os.getcwd())
sys.path.append(parent_dir)
import data_prep as dp
import ccf
from radialVelocityCalc import getHeaderValues
from PyAstronomy import pyasl
import numpy as np

#This should allow the files in the current git library to be used as inputs. So you should be able to run this test from your computer as well
testing_dir = os.getcwd()

#First, I am testing on two spectra that the LCO pipeline succeeded on. The LCO results for these files are also in the git repo
observedFile = os.path.join(testing_dir, "testingGoodObserved.fits.fz")
templateFile = os.path.join(testing_dir, "testingGoodTemplate.fits.fz")
```

**Test 3: Interpolation**

This test is focused on the interpolation part of the code, and it is intended to see if the wavelengths shape is signifcantly changed during the interpolation. For this test, I will use an easier, more well-behaved function in the sin function instead of the spectra. I will interpolate values of the sin function that are on a grid. Then, I will calculate the residuals of the new, interpolation values vs true values of the sin function on the new grid. Ideally, the residuals should be low and the plot of the interpolated values should match the true values.


```python
import matplotlib.pyplot as plt
x = np.linspace(1, 3*3.14, 2000)
y = np.sin(x)

xInterpolated, yInterpolated, _ = dp.interpolation(x, y, x, y) #we only have one set of values
yTrue = np.sin(xInterpolated)
plt.plot(xInterpolated, yTrue, label='True')
plt.plot(xInterpolated, yInterpolated, '--', label='Interpolated')
plt.show()

residuals = yInterpolated - yTrue
print(str(np.max(residuals)))
```

Firstly, we can see on the plot that the interpolated y values and true y values are basically on the same line. Also, the maximum residual value is very low. These resutls give me confidence that the interpolation function is behaving well and not affecting the wavelengths data; the spectra's shapes stay largely the same as before interpolation.
